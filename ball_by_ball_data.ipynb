# **Batting performance during powerplay**

```python
#phase of innings
powerplay_overs=range(0,7)
middle_overs=range(7,16)
death_overs=range(17,21)

#feature engg
#create column innings_phase to idesntify phase of each delivery
def categorise_innings_phase(overs):
  if overs in powerplay_overs:
    return 'Powerplay'
  elif overs in middle_overs:
    return 'Middle Overs'
  elif overs in death_overs:
    return 'Death overs'
  else:
    return 'other'

data['innings_phase']=data['overs'].apply(categorise_innings_phase)
data.info()
```

```python
phase_group=data.groupby('innings_phase')

# calculate various batting performance metric for each phase
phase_stats=phase_group.agg({'total_run':'sum','batsman_run':'sum','ballnumber':'count'})
phase_stats
```

```python
powerplay_overs= range(0,7)
middle_overs=range(7,16)
death_overs=range(16,21)

#filter for powerplay, middle overs and death overs
powerplay_df=data[data['overs'].isin(powerplay_overs)]
middleovers_df=data[data['overs'].isin(middle_overs)]
deathovers_df=data[data['overs'].isin(death_overs)]

#calculate various batting performance metrics for each phase
powerplay_stats=powerplay_df.agg({'total_run':'sum','batsman_run':'sum',"ballnumber":'count'})
middle_over_stats=middleovers_df.agg({'total_run':'sum','batsman_run':'sum',"ballnumber":'count'})
deathover_stats=deathovers_df.agg({'total_run':'sum','batsman_run':'sum',"ballnumber":'count'})

powerplay_stats['strike_rate']=(powerplay_stats['batsman_run']/powerplay_stats['ballnumber'])*100
middle_over_stats['strike_rate']=(middle_over_stats['batsman_run']/middle_over_stats['ballnumber'])*100
deathover_stats['strike_rate']=(deathover_stats['batsman_run']/deathover_stats['ballnumber'])*100

powerplay_stats
#middle_over_stats
#deathover_stats

```

# **statistics to dataframe format and transpose**

```python
power_play_stats_df=pd.DataFrame(powerplay_stats)
power_play_stats_df
```

```python
power_play_stats_df=pd.DataFrame(powerplay_stats).transpose()
power_play_stats_df.columns=['total_run'," kohli_runs",'balls_faced',"strike_rate"]

middle_overs_stats_df=pd.DataFrame(middle_over_stats).transpose()
middle_overs_stats_df.columns=['total_run'," kohli_runs",'balls_faced',"strike_rate"]

death_overs_stats_df=pd.DataFrame(deathover_stats).transpose()
death_overs_stats_df.columns=['total_run'," kohli_runs",'balls_faced',"strike_rate"]

death_overs_stats_df
```

```python

#display results!!
print('POWER PLAY STATISTICS!')
print(power_play_stats_df)
print("\n MIDDLE OVERS STATISTICS")
print(middle_overs_stats_df)
print("\n DEATH OVERS STATISTICS")
print(death_overs_stats_df)
```

### Converting float columns to integers


```python
power_play_stats_df[['total_run'," kohli_runs",'balls_faced',"strike_rate"]]=power_play_stats_df[['total_run'," kohli_runs",'balls_faced',"strike_rate"]].astype(int)
middle_overs_stats_df[['total_run'," kohli_runs",'balls_faced',"strike_rate"]]=middle_overs_stats_df[['total_run'," kohli_runs",'balls_faced',"strike_rate"]].astype(int)
death_overs_stats_df[['total_run'," kohli_runs",'balls_faced',"strike_rate"]]=death_overs_stats_df[['total_run'," kohli_runs",'balls_faced',"strike_rate"]].astype(int)

#display results!!
print('POWER PLAY STATISTICS!')
print(power_play_stats_df)
print("\n MIDDLE OVERS STATISTICS")
print(middle_overs_stats_df)
print("\n DEATH OVERS STATISTICS")
print(death_overs_stats_df)
```
