concat()

used to concatenate multiple dataframes

```python
df1= pd.DataFrame({'Player':['SuryaKumar Yadav','Rohit Sharma','Jasprit Bumrah'],'Team': ['MI','MI','MI'],'Runs':[11867,9115,0],'Wickets':[0,0,97]})
```

```python
df2=pd.DataFrame({'Player':['Virat Kohli','Faf Duplesis','Maxwell'],'Team':['RCB','RCB','RCB'],'Runs':[10115,9115,10767]})
```

```python
df3=pd.DataFrame({'Player':['Ashwin','Siraj','Varun Chakravarthy'],'Team':['RR','RCB','KKR'],'Wickets':[101,91,107]})
```

now concatenating the dataframes

```python
concatenated_df=pd.concat([df1,df2,df3])
print(concatenated_df)
```

merge()

combine 2 or more data frames based on a common column

```python
df4=pd.DataFrame({'Player':['MS Dhoni','Rahane','Ruturaj'],'Team':['CSK','CSK','CSK'],'Runs':[11867,9115,27]})
```

```python
df5=pd.DataFrame({'Player':['MS Dhoni','Rahane','Kohli'],'Team':['CSK','CSK','RCB'],'Average':[59.33,48.55,54.17]})
```

```python
merged_df=pd.merge(df4,df5,on='Player')
print(merged_df)
```

InnerJoin()

Combines only the maching rows frok two DataFrames based on common column.


```python
innerjoin_df=pd.merge(df4,df5, on='Player',how='inner')
print(innerjoin_df)
```

outer join

Retuns all rows from both data frames and fills missing values wth NaN for any non matching rows

```python
print(df4)
print('\n')
print(df5)
print('\n')
outerjoindf= pd.merge(df4,df5,on='Player',how='outer')
outerjoindf
```

leftjoin()

Returns all rows from the left dataframe and the matching rows from the right dataframe

If there is no matching row in the right dataframe , the missing values are filled wih NaN

```python
print(df4)
print('\n')
print(df5)
print('\n')
leftjoindf=pd.merge(df4,df5,on='Player',how='left')
print(leftjoindf)
```

right join

returns all the rows from the right dataframe and the matching rows from the left dataframe.

If there is no matching row in the left dataframe, missing values are filled with NaN.

```python
print(df4)
print('\n')
print(df5)
print('\n')
rightjoindf=pd.merge(df4,df5, on='Player',how='right')
rightjoindf
```
